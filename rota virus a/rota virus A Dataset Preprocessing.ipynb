```python
import os
import re
import nltk
import sklearn
import pandas as pd
from tqdm.notebook import tqdm, trange
import random
```

```python
#importing the dataset
dataset = pd.read_csv("rota_dataset.csv")
```

```python
#splitting into train test set
from sklearn.model_selection import train_test_split
X_train,X_test,y_train, y_test = train_test_split(dataset['SEQUENCE'], dataset['HOST'], test_size=0.2,random_state=109,stratify=dataset['HOST'])
```

```python
trainArrayX=[]
testArrayX=[]

trainArrayY=[]
testArrayY=[]
for i in X_train:
    trainArrayX.append(i)
for i in X_test:
    testArrayX.append(i)

    
for i in y_train:
    trainArrayY.append(i)
for i in y_test:
    testArrayY.append(i)
```

```python
data={}
data['SEQUENCE']=[]
data['HOST']=[]
df_train=pd.DataFrame(data)
df_test=pd.DataFrame(data)

#converting into dataset from list
for i in trange(len(trainArrayX)):
    new={}
    new['SEQUENCE'] = trainArrayX[i]
    new['HOST'] = trainArrayY[i]
    df_train = pd.concat([df_train, pd.DataFrame([new])], ignore_index=True)

#converting into dataset from list
for i in trange(len(testArrayX)):
    new={}
    new['SEQUENCE'] = testArrayX[i]
    new['HOST'] = testArrayY[i]
    df_test = pd.concat([df_test, pd.DataFrame([new])], ignore_index=True)
```

```python
df_train.to_csv('rota_virus_training_dataset_Without_Preprocessing.csv', index=False)
df_test.to_csv('rota_virus_test_dataset_Without_Preprocessing.csv', index=False)
```

```python
data={}
data['SEQUENCE']=[]
data['HOST']=[]
dataset_train_resized=pd.DataFrame(data)
dataset_test_resized=pd.DataFrame(data)
```

```python
#resizing dataset to length 400 bases
def resizeDataset(sourceDataset, destinationDataset):
    for i in trange(len(sourceDataset)):
        new={}
        if(len(sourceDataset['SEQUENCE'][i]) >=400):
            new['SEQUENCE'] = sourceDataset['SEQUENCE'][i][:400]
        else:
            holdGenome = sourceDataset['SEQUENCE'][i]
            while(len(holdGenome) <400):
                holdGenome = holdGenome+sourceDataset['SEQUENCE'][i]
            new['SEQUENCE'] = holdGenome[:400]
    
        new['HOST'] = sourceDataset['HOST'][i]
        destinationDataset = pd.concat([destinationDataset, pd.DataFrame([new])], ignore_index=True)
    return destinationDataset
```

```python
dataset_train_resized = resizeDataset(df_train,dataset_train_resized)
dataset_test_resized = resizeDataset(df_test,dataset_test_resized)
```

```python
#replacing unknown charaacters with N
def replaceUnknownCharacter(dataset_resized):
    data={}
    data['SEQUENCE']=[]
    data['HOST']=[]
    dataset_ACGTN=pd.DataFrame(data)
    #replacing charachters which is not ACGTN with N
    for i in trange(len(dataset_resized)):
        new={}
        holdGenome=''
        for x in dataset_resized['SEQUENCE'][i]:
            if x in ['A', 'G', 'C', 'T', 'N']:
                holdGenome = holdGenome+x
            else:
                holdGenome = holdGenome+ 'N'
        new['SEQUENCE'] = holdGenome
    
        new['HOST'] = dataset_resized['HOST'][i]
        dataset_ACGTN = pd.concat([dataset_ACGTN, pd.DataFrame([new])], ignore_index=True)
    return dataset_ACGTN
```

```python
dataset_train_resized = replaceUnknownCharacter(dataset_train_resized)
dataset_test_resized = replaceUnknownCharacter(dataset_test_resized)
```

```python
dataset_train_resized.to_csv('rota_virus_training_dataset_preproccessed.csv', index=False)
dataset_test_resized.to_csv('rota_virus_test_dataset_preproccessed.csv', index=False)
```
