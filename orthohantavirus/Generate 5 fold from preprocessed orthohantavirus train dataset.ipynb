```python
import os
import re
import nltk
import sklearn
import pandas as pd
from tqdm.notebook import tqdm, trange
import random
from sklearn.model_selection import StratifiedKFold
```

```python
dataset = pd.read_csv("orthohanta_virus_training_dataset_preproccessed.csv")
kfold = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
fold = 1
for train, val in kfold.split(dataset['SEQUENCE'], dataset['HOST']):

    data={}
    data['SEQUENCE']=[]
    data['HOST']=[]
    df_train=pd.DataFrame(data)
    df_val=pd.DataFrame(data)
    for i in trange(len(train)):
        new={}
        new['SEQUENCE'] = dataset['SEQUENCE'][train[i]]
        new['HOST'] = dataset['HOST'][train[i]]
        df_train = pd.concat([df_train, pd.DataFrame([new])], ignore_index=True)

    for i in trange(len(val)):
        new={}
        new['SEQUENCE'] = dataset['SEQUENCE'][val[i]]
        new['HOST'] = dataset['HOST'][val[i]]
        df_val = pd.concat([df_val, pd.DataFrame([new])], ignore_index=True)
    df_train.to_csv('orthohanta_virus_Fold_'+str(fold)+'_training_dataset_preproccessed.csv', index=False)
    df_val.to_csv('orthohanta_virus_Fold_'+str(fold)+'_validation_dataset_preproccessed.csv', index=False)
    fold=fold+1
```
