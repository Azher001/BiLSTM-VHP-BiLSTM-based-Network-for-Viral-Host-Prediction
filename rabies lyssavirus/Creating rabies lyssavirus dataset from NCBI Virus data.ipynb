```python
import os
import re
import nltk
import sklearn
import pandas as pd
from tqdm.notebook import tqdm, trange
from Bio import SeqIO
```

```python
dataset = pd.read_csv("sequences.csv")
dataset.dropna(inplace=True)
dataset=dataset.reset_index(drop=True)
```

```python
#Creating a dictionary to hold host count

dict={}
for i in trange(len(dataset)):
    if dict.get(dataset['Host'][i]) != None:
        dict[dataset['Host'][i]] = dict[dataset['Host'][i]]+1;
    else:
        dict[dataset['Host'][i]] = 1;
```

```python
dataX={}
dataX['Accession']=[]
dataX['Host']=[]
dataset_100=pd.DataFrame(dataX)
```

```python
#keeping the accession and host information for the hosts with more than 100 count

for i in trange(len(dataset)):
    if dict.get(dataset['Host'][i]) > 100:
        new={}
        new['Accession'] = dataset['Accession'][i]
        new['Host'] = dataset['Host'][i]
        dataset_100 = pd.concat([dataset_100, pd.DataFrame([new])], ignore_index=True)
```

```python
records = list(SeqIO.parse("sequences.fasta", "fasta"))
```

```python
dataX={}
dataX['SEQUENCE']=[]
dataX['HOST']=[]
df_rabies=pd.DataFrame(dataX)
```

```python
#creating the dataset with sequence and host information by matching the accession number
for index in trange(len(dataset_100)):
   
    if dataset_100['Host'][index] != '':
        accession = dataset_100['Accession'][index]
        for i in range(len(records)):
            recordsAccession = records[i].id
            if accession == recordsAccession:
                new={}
                new['SEQUENCE'] = str(records[i].seq)
                new['HOST'] = dataset_100['Host'][index]
                df_rabies = pd.concat([df_rabies, pd.DataFrame([new])], ignore_index=True)
                break
            else:
                continue
```

```python
df_rabies.to_csv('rabies_dataset.csv', index=False)
```
